```python
import pandas as pd


df1 = pd.read_csv("mexico_real_estate.csv")
df1.head(10)
```

It looks like there are a couple of problems in this DataFrame that you need to solve. First, there are many rows with NaN values in the "lat" and "lon" columns. Second, the data type for the "price_usd" column is object when it should be float.

Task 1.2.3: Clean df1 by dropping rows with NaN values. Then remove the "$" and "," characters from "price_usd" and recast the values in the column as floats

```python
#To drop rows with missing values from a DataFrame in pandas, use the dropna() method.
# By default, df.dropna() removes any row that contains at least one missing value (NaN or None) and returns a new DataFrame.
df1 = df1.dropna()

df1
```

```python
#To replace string characters in a pandas column, use the .str.replace() method.
# This allows you to target a specific column and swap characters or text patterns.

df1['price_usd'] = df1['price_usd'].str.replace('$', '').str.replace(',', '').astype(float)

df1
```

Clean df3
 We're now on the final DataFrame. Use the same shape, info and head commands to inspect the df3. Do you see any familiar issues?

You'll notice that we still have NaN values, but there are two new problems:

```python
print(df3.shape)
print(df3.head(10))
print(df3.dtypes)
print(df3.isnull().sum())

df3
```

Task 1.2.5: Drop rows with NaN values in df3. Then use the split method to create two new columns from "lat-lon" named "lat" and "lon", respectively.

```python
#To split strings in one column and create new columns from the result in pandas, use the .str.split() method with expand=True.
#df["col"].str.split(sep) splits each string by sep (default is whitespace).
#expand=True turns the list of split parts into separate columns (returns a DataFrame).
#Assigning to df[["new1", "new2", ...]] creates those new columns in one step.
#n=1 means “split at most once”, so you get exactly two columns even if there are more commas
df3[["lat","lon"]] = df3["lat-lon"].str.split(",",n = 1, expand=True)

df3 = df3.drop(columns=["lat-lon"])

df3
```

```python
df3 = df3.dropna()
df3['price_usd'] = df3['price_usd'].str.replace('$', '').str.replace(',', '').astype(float)

df3
```

Task 1.2.6: Use the split method again, this time to extract the state for every house. (Note that the state name always appears after "México|" in each string.) Use this information to create a "state" column. Finally, drop the "place_with_parent_names" and "lat-lon" columns from the DataFrame

```python
#Creating just one new column from a split
#If you only need one part of the split (e.g., the part after a delimiter), you can index the list

df3['state'] = df3['place_with_parent_names'].str.split("|").str[0]

df3 = df3.drop(columns=["place_with_parent_names"])

df3
```

Concatenate DataFrames
You have three clean DataFrames, and now it's time to combine them into a single DataFrame so that you can conduct your analysis.

Task 1.2.7: Use pd.concat to stack df1, df2, and df3 vertically into a new DataFrame named df. After dropping rows with missing values in each cleaned frame, the result should have 1,650 rows and 6 columns: "property_type", "state", "lat", "lon", "area_m2", and "price_usd".


```python
# Stack the cleaned DataFrames by rows instead of side by side.
# ignore_index=True resets the index so the final DataFrame is clean and sequential.
df = pd.concat([df1, df2, df3], ignore_index=True)

df
```
